## 3. Generate Sample Data

```python
# Generate some sample data for analysis
def generate_sample_data():
    """Generate sample data for visualization"""
    # Sample query performance data
    np.random.seed(42)
    
    # Query types and performance
    query_types = ['SELECT', 'INSERT', 'UPDATE', 'DELETE', 'JOIN']
    query_counts = np.random.randint(100, 1000, size=len(query_types))
    query_times = np.random.exponential(scale=100, size=len(query_types))
    
    # Create DataFrame
    query_df = pd.DataFrame({
        'Query Type': query_types,
        'Count': query_counts,
        'Avg Time (ms)': query_times
    })
    
    # Table access patterns
    tables = ['customers', 'orders', 'products', 'order_items', 'logs', 'metrics']
    access_counts = np.random.randint(1000, 10000, size=len(tables))
    
    table_df = pd.DataFrame({
        'Table': tables,
        'Access Count': access_counts
    })
    
    return query_df, table_df

# Generate data
query_df, table_df = generate_sample_data()

print("Sample Data Generated:")
print("\nQuery Statistics:")
print(query_df)
print("\nTable Access Patterns:")
print(table_df)
```

## 4. Visualize Data

```python
# Create visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Query type distribution
axes[0, 0].bar(query_df['Query Type'], query_df['Count'], color='skyblue', edgecolor='black')
axes[0, 0].set_title('Query Type Distribution')
axes[0, 0].set_xlabel('Query Type')
axes[0, 0].set_ylabel('Count')
axes[0, 0].tick_params(axis='x', rotation=45)

# 2. Query performance
axes[0, 1].bar(query_df['Query Type'], query_df['Avg Time (ms)'], color='lightcoral', edgecolor='black')
axes[0, 1].set_title('Average Query Execution Time')
axes[0, 1].set_xlabel('Query Type')
axes[0, 1].set_ylabel('Time (ms)')
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Table access patterns
axes[1, 0].barh(table_df['Table'], table_df['Access Count'], color='lightgreen', edgecolor='black')
axes[1, 0].set_title('Table Access Patterns')
axes[1, 0].set_xlabel('Access Count')
axes[1, 0].set_ylabel('Table')

# 4. Query count vs time scatter
axes[1, 1].scatter(query_df['Count'], query_df['Avg Time (ms)'], s=100, alpha=0.6, edgecolor='black')
axes[1, 1].set_title('Query Count vs Execution Time')
axes[1, 1].set_xlabel('Count')
axes[1, 1].set_ylabel('Avg Time (ms)')
for i, txt in enumerate(query_df['Query Type']):
    axes[1, 1].annotate(txt, (query_df['Count'][i], query_df['Avg Time (ms)'][i]))

plt.tight_layout()
plt.show()
```

## 5. Test Feature Extraction

```python
try:
    from feature_extractor.pipeline import FeatureExtractor, create_feature_extractor
    
    # Create a feature extractor
    extractor = create_feature_extractor({
        'normalize': True,
        'include_temporal': True,
        'feature_dim': 128  # Smaller for testing
    })
    
    # Create sample feature data
    workload_example = {
        'query_type_distribution': {'select': 0.7, 'insert': 0.2, 'update': 0.1},
        'performance_metrics': {
            'latency_percentiles': {'p50': 10, 'p95': 50, 'p99': 100},
            'throughput_avg': 1000,
            'cache_hit_ratio_avg': 0.95
        }
    }
    
    config_example = {
        'summary': {
            'total_indexes': 50,
            'total_tables': 10
        }
    }
    
    performance_example = {
        'mean_latency': 25.5,
        'total_throughput': 1000
    }
    
    # Extract features
    features = extractor.extract_state_vector(workload_example, config_example, performance_example)
    
    print("Feature extraction successful!")
    print(f"Feature vector shape: {features.shape}")
    print(f"Feature vector length: {len(features)}")
    print(f"\nFirst 10 features: {features[:10]}")
    
    # Show feature statistics
    print(f"\nFeature Statistics:")
    print(f"  Min: {features.min():.4f}")
    print(f"  Max: {features.max():.4f}")
    print(f"  Mean: {features.mean():.4f}")
    print(f"  Std: {features.std():.4f}")
    
except ImportError as e:
    print(f"Cannot import feature extractor: {e}")
    print("Make sure the src directory is in your Python path.")
except Exception as e:
    print(f"Error in feature extraction: {e}")
```
